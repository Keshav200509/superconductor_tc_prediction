```python
import pandas as pd
df = pd.read_csv("../data/train.csv")
print(df.shape)
df.head()
```

```python
df.info()
df.describe()
```

```python
df.isnull().sum().sum()
```

```python
import matplotlib.pyplot as plt

plt.hist(df['critical_temp'], bins=50)
plt.title("Distribution of Tc")
plt.xlabel("Critical Temperature")
plt.ylabel("Frequency")
plt.show()
```

```python
import seaborn as sns

corr = df.corr()

# Top 10 features correlated with Tc
top_corr = corr['critical_temp'].abs().sort_values(ascending=False)[1:11]

sns.barplot(x=top_corr.values, y=top_corr.index)
plt.title("Top Features Correlated with Tc")
plt.show()
```
